##### Independent feature is balanced

##### Now we separate the dependent and independant features.

```python
x = data.iloc[:,1:]
y = data.iloc[:,0]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from  sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred_train = model.predict(x_train)
y_pred_test = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Train Accuracy:", accuracy_score(y_train, y_pred_train))
print("Test Accuracy:", accuracy_score(y_test, y_pred_test))
```

```python
print("Training Classification Report:\n", classification_report(y_train, y_pred_train))
```

```python
print("Testing Classification Report:\n", classification_report(y_test, y_pred_test))
```

```python
print("Training Confusion Matrix:\n", confusion_matrix(y_train, y_pred_train))
```

```python
print("Test Confusion Matrix:\n", confusion_matrix(y_test, y_pred_test))  
```

```python
from sklearn.metrics import confusion_matrix

def get_confusion_matrix_df(cm_func, y_true, y_pred):
    cm = cm_func(y_true, y_pred)
    return pd.DataFrame(cm, 
                        index=['Actual True', 'Actual False'], 
                        columns=['Predicted True', 'Predicted False'])

# Use function directly here without redefining it
cm_df = get_confusion_matrix_df(confusion_matrix, y_test, y_pred_test)
cm_df
```

```python
print("Training Accuracy:", accuracy_score(y_train, y_pred_train))
print("Test Accuracy:", accuracy_score(y_test, y_pred_test))
```

```python
cm_df = get_confusion_matrix_df(confusion_matrix, y_test, y_pred_test)
print("Testing Confusion Matrix:\n", cm_df)
```

```python
print("Test Classification Report:\n", classification_report(y_test, y_pred_test))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
y_pred_proba = model.predict_proba(x_test)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)
print("ROC AUC Score:", roc_auc)
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label='ROC Curve (AUC = {:.2f})'.format(roc_auc))
plt.plot([0, 1], [0, 1], 'k--', label='Random Guessing')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.show()
```
